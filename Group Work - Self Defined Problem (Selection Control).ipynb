```python
# Program to calculate delivery cost based on weight of the item

# Get weight of the item 
weight = int(input("Enter item weight in kg: "))

# Check conditions to find out total delivery charge
if weight <= 2:
    cost = 3
elif weight >2 and weight <= 5:
    cost = 5
elif weight >5 and weight <=10:
    cost = 8
elif weight >10 and weight <= 20:
    cost = 12
else:
    cost = 18

# Output result
print(f"Total delivery cost: £{cost}")
```
