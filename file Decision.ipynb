if_else statement

```python
#ask user to enter the activity to be performed
print("please enter the activity to be performed!")
activity= input("please enter the activity to be performed: ")
#determine if the activity is calculation or not
if activity == "calculate":
    print("performing calculations...")
else:    
    print("performing activity")
print("activity completed.")
```

if_elif_else statement

```python
#ask user to enter the direction to paint
print("towards which direction should i paint?(up, down, left, right)")
direction = input("towards which direction should i paint (up, down, left, right): ")
#determine the message to display based on the direction entered by the user
if direction == "up":
    print("i am painting in the upward direction...")
elif direction == "down":
    print("i am painting in the downward direction...")
elif direction == "left":
    print("i am painting in the leftward direction...")
elif direction == "right":
    print("i am painting in the rightward direction...")
else:
    print("Invalid direction. Please enter a valid direction.")
```

modulo_operator

```python
#ask user to enter a whole number
print("enter a whole number.")
number = int(input("please enter a whole number: "))
#determine if the number is even or odd
if number % 2 == 0:
    print(f"{number} is an even number.")
else:
    print(f"{number} is an odd number.")    
```
